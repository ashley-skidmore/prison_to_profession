# Cleaning

## Column Names

Using the data dictionary provided for this dataset, I will start by changing column names to descriptive ones.

```python
df = df.rename(columns={"SERIALNO": "serial_number","TYPEHUGQ": "group_quarters_type"})

df.head()
```

## Mapping Coded Values
Using ACS Data Dictionary

This section converts coded values into descriptive labels.

```python
group_quarters_map = {
    1: "Housing unit",
    2: "Institutional group quarters",
    3: "Noninstitutional group quarters"
}

df["group_quarters_type"] = df["group_quarters_type"].map(group_quarters_map)

df.head()
```

## Rechecking Data Types

```python
df.dtypes
```

# Converting to SQLite Database Table

```python
# Connect to existing SQLite database
conn = sqlite3.connect("../data/cleaned_data.sqlite")

# Write the DataFrame `df` to the database as a new table called "acsh_data"
df.to_sql(
    name="acsh_data",       # name of the new table
    con=conn,               # database connection
    if_exists="replace",    # overwrite if table already exists
    index=False             # don't include the index as a column
)

# Close the connection
conn.close()

print("Table 'acsh_data' successfully added to cleaned_data.sqlite.")

```
